`Show3D` scrubs a 3D volume slice by slice: a multislice ptychographic object phase, a tomographic reconstruction, a focal series. Drag the slider (or use the arrow keys) to move through depth. Exported with `offline=True`, so the scrub works in this static page with no kernel.

```{note}
The widget below is exported with `offline=True`, which bakes the data as a **uint8 quantized** stack - kernel-less and 4x smaller than float32, visually identical after the colormap. This is the same packing `export_html(quantized=True)` writes for sharing.
```

````python
import numpy as np

# Synthetic ptychographic-object-phase volume: atom columns that drift across
# depth, so scrubbing through z shows the lattice shift slice by slice.
depth, size = 64, 256
row, col = np.mgrid[0:size, 0:size]
spacing = 20
sites = np.arange(spacing // 2, size, spacing)
volume = np.zeros((depth, size, size), dtype=np.float32)
for z in range(depth):
    shift = 6.0 * np.sin(2 * np.pi * z / depth)  # columns sway with depth
    for r0 in sites:
        for c0 in sites:
            volume[z] += np.exp(-(((row - r0 - shift) ** 2 + (col - c0) ** 2)) / (2 * 2.5 ** 2))
volume.shape, volume.dtype
````

## Scrub the volume

Wrap the stack in a quantem `Dataset3d` so the volume carries depth and lateral calibration. `Show3D` reads the dataset metadata automatically; no widget-level sampling argument is needed. Opens on the middle slice at 30 fps. The atom columns sway as you move through depth.

````python
from quantem.core.datastructures import Dataset3d
from quantem.widget import Show3D

volume_dataset = Dataset3d.from_array(
    volume,
    sampling=(0.6, 0.18, 0.18),
    units=("nm", "nm", "nm"),
    name="Synthetic object phase",
)

Show3D(volume_dataset, offline=True)
````
