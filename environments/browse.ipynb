# Browse a 4D-STEM scan

Set `SCAN` to your file, then run the cells. The scan stays lossless and compressed on the GPU: nothing is binned or cropped.

```python
from pathlib import Path
import quantem.widget as qw
from quantem.gpu import io
from quantem.widget import Show4DSTEM
SCAN = Path(r"D:\data\scan_master.h5")  # an Arina master (its data files beside it) or a .qem file
```

```python
qw.profile()
```

```python
data = io.load(SCAN)
Show4DSTEM(data)
```
