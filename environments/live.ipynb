# Live: the detector folder, and direct ptychography below

Each scan appears in the viewer once its files have finished arriving. The panel below fits its aberrations through fifth order on the central 256 x 256 scan region and shows the whole scan's phase; drag the dashed fit box on it to fit somewhere else.

```python
from pathlib import Path
import quantem.widget as qw
from quantem.widget import Show4DSTEM, ShowPtycho
FOLDER = Path(r"D:\data\today")  # the folder the detector writes its scans into
```

```python
qw.profile()
```

```python
viewer = Show4DSTEM.live(
    FOLDER,
    latest=1,  # scans kept on the GPU; a 4 GB GPU holds one 512 x 512 scan
    semiangle_mrad=30,  # the probe's convergence semiangle; the detector files do not record it
)
```

```python
ptycho = ShowPtycho.live(viewer)  # aberrations on the central 256 x 256, the whole phase
```
