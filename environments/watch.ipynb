# Watch the detector folder, then direct ptychography

Set `FOLDER` to the folder the detector writes into. Each scan appears once its files have finished arriving. The last cell fits the aberrations through fifth order on the scan shown and reconstructs its phase; run it again for the next scan, or press **Direct ptycho** in the viewer.

```python
from pathlib import Path
import quantem.widget as qw
from quantem.widget import Show4DSTEM
FOLDER = Path(r"D:\data\today")  # the folder the detector writes its scans into
```

```python
qw.profile()
```

```python
viewer = Show4DSTEM.watch(
    FOLDER,
    latest=1,  # scans kept on the card; a 4 GB card holds one 512 x 512 scan
)
viewer
```

```python
phase = viewer.direct_phase()  # aberrations through fifth order on the scan shown, then its whole phase
```
