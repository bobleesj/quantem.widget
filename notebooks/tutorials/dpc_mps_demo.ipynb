# 4D-STEM on MacBook (MPS) - virtual images + DPC, no binning

One container, one API. Apple Metal on MacBook, torch on CUDA - no backend awareness,
no binning, no quantem dependency. Bit-exact vs quantem.live (parity-tested).


## Environment


```python
from quantem.gpu import detector, dpc
from quantem.gpu.io import load
from quantem.widget import Show2D, device_info
device_info()                       # version, date, GPU / MacBook chip
```

## Load (no bin) -> one encoded GPU acquisition


```python
loaded = load("path/to/your_master.h5", verbose=True)
loaded                              # Dataset4dstemGPU(shape=(512, 512, 192, 192), dtype=uint16, representation='encoded')
```

## Virtual images  (auto probe-fit: BF / ABF / ADF / HAADF / DF / DP)


```python
Show2D(detector.virtual(loaded, "BF"))
```

```python
Show2D(detector.virtual(loaded, "ADF"))
```

## DPC / iDPC  (CoM -> rotation -> integrated phase)


```python
result = dpc.run(loaded, verbose=True)
Show2D(result.phase)                # iDPC - projected potential / atoms
```

```python
Show2D(result.com_col)              # horizontal CoM - raw DPC field
```
