# Browse a scan and run direct ptychography

Set `SCAN` to your file, then run the cells. The viewer shows the scan, lossless and compressed on the GPU. The panel below fits its aberrations through fifth order on the central 256 x 256 scan region (128 x 128 on a GPU of under 6 GB, about twice as fast there) and shows the whole scan's phase; drag the dashed fit box on it to fit somewhere else. Hover a kept fit to look at it again, click it to fit there again.

```python
from pathlib import Path
import quantem.widget as qw
from quantem.widget import Show4DSTEM, ShowPtycho
SCAN = Path(r"D:\data\scan_master.h5")  # an Arina master (its data files beside it) or a .qem file
```

```python
qw.profile()
```

```python
viewer = Show4DSTEM.live(
    SCAN,
    latest=1,
    semiangle_mrad=30,  # the probe's convergence semiangle; the detector files do not record it
)
```

```python
ptycho = ShowPtycho.live(viewer)  # aberrations on the central 256 x 256 (128 x 128 on a GPU under 6 GB), the whole phase
```
